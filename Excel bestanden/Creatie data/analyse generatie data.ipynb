```python
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency
import numpy as np

bloeddruk_verdeling = {
    '<120': (22, 32),
    '120-140': (31, 11),
    '140-180': (15, 7),
    '>180': (0, 0)
}

categorieen = list(bloeddruk_verdeling.keys())

# splits data
dataset1 = np.array([v[0] for v in bloeddruk_verdeling.values()])
dataset2 = np.array([v[1] for v in bloeddruk_verdeling.values()])

# percentages
perc1 = dataset1 / dataset1.sum() * 100
perc2 = dataset2 / dataset2.sum() * 100

x = np.arange(len(categorieen))
breedte = 0.35

fig, ax = plt.subplots(layout='constrained')

bars1 = ax.bar(x - breedte/2, perc1, breedte, label='Hartstikke gezond (n=68)')
bars2 = ax.bar(x + breedte/2, perc2, breedte, label='Gegenereerde data (n=50)')

ax.bar_label(bars1, fmt='%.1f%%', padding=3)
ax.bar_label(bars2, fmt='%.1f%%', padding=3)

ax.set_ylabel('Percentage (%)')
ax.set_title('Vergelijking bloeddrukverdeling')
ax.set_xticks(x)
ax.set_xticklabels(categorieen)
ax.set_ylim(0, 100)
ax.legend()

#plt.savefig("bloeddruk_verdeling.png", dpi=300, bbox_inches="tight")
plt.show()

data = np.array([
    [22, 32],
    [31, 11],
    [15, 7]
])

chi2, p, dof, expected = chi2_contingency(data)
print(p)
```
